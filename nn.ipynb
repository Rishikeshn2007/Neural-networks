Building nural network from scratch to predict the output of a XOR gate with 2 inputs 1 output and 1 hidden layer with 3 neurons:

```python
import numpy as np
```

Initializing random values for all weights and bias of the network.

```python
print("Layer 1 weights:")
w_layer1=np.random.rand(2,3)
b_layer1=np.random.rand(1,3)
print(w_layer1)
print("Layer 1 bias:")
print(b_layer1)

w_layer2=np.random.rand(3,1)
b_layer2=np.random.rand(1)
print("Layer 2 weights:")
print(w_layer2)
print("Layer 2 bias:")
print(b_layer2)
```

Function to give back thw summation od weights and bias to the signum function.

```python
def summation(inputs, weights, bias):
    Z=np.dot(inputs,weights)+bias
    return Z
```

Function to apply signum function over a matrix Z(layer) and return back the activation matrix A:

```python
def signum(Z_layer):
    A=np.sign(Z_layer)
    return A
```

Sample data:

```python
X=np.array([[1,1],[1,0],[0,1],[0,0]])
Y=np.array([0,1,1,0])
print("Input data:")
print(X)
print("Actual result:")
print(Y)
```

Calculationg summations and activations for each of these values and data set and looking up the prediction:

```python
print("Layer 1 weighted sum:")
Z1=summation(inputs=X[0:1],weights=w_layer1,bias=b_layer1)
print(Z1)
print("Activations of layer 1:")
H=signum(Z1)
print(Z1)
print("Layer 2 weighted sum:")
Z2=summation(inputs=H,weights=w_layer2,bias=b_layer2)
print(Z2)
print("Activations of layer 2 (The final output):")
O=signum(Z2)
print(O)
```

Gradient desent and backpropagarion:

```python
learnig_rate=0.1
error=Y[0:1]-O
print(f"Error:{error}")

w_layer2 += learnig_rate * np.dot(H.T, error)
print(w_layer2)
```
